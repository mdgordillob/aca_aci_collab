## 1. Find the exact path

## 2. Walk the subtree

Edit `startpath` below to match what step 1 showed, then run. Prints every folder with a file count and a sample of filenames (so naming conventions like `era5_tmp_1985.grib` are visible, not just folder names).

```python
import os

startpath = "/content/drive/MyDrive/2. Datos"  # <-- edit to match step 1
MAX_FILES_SHOWN = 8

assert os.path.isdir(startpath), (
    f"{startpath!r} not found -- os.walk() silently prints nothing on a bad "
    "path instead of erroring, so check this first. Compare against the exact "
    "names printed by step 1 (repr() there would reveal accents/spaces)."
)

def list_tree(startpath, max_files=MAX_FILES_SHOWN):
    for root, dirs, files in os.walk(startpath):
        level = root.replace(startpath, '').count(os.sep)
        indent = '  ' * level
        print(f"{indent}{os.path.basename(root)}/  ({len(files)} files)")
        subindent = '  ' * (level + 1)
        for f in sorted(files)[:max_files]:
            print(f"{subindent}{f}")
        if len(files) > max_files:
            print(f"{subindent}... ({len(files) - max_files} more)")

list_tree(startpath)
```
